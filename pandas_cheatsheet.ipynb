```python
df = pd.DataFrame(
    {"id": [1, 2, 3, 4, 5, 6], "raw_grade": ["a", "b", "b", "a", "a", "e"]}
)
df
```

```python
df["grade"] = df["raw_grade"].astype("category")
df["grade"]
```

```python
ts = pd.Series(np.random.randn(1000), index=pd.date_range("1/1/2000", periods=1000))
ts = ts.cumsum()
ts.plot();
```

ts = ts.cumsum(): Tính tổng tích lũy


```python
df = pd.DataFrame(
    np.random.randn(1000, 4), index=ts.index, columns=["A", "B", "C", "D"]
)

df = df.cumsum()
plt.figure();
df.plot();
plt.legend(loc='best');
```

Tạo ma trận:

1000 hàng,
4 cột

chứa số ngẫu nhiên theo phân phối chuẩn.

index=ts.index: dùng index từ thời gian ts

; -> để notebook không in output thừa.

plt.legend(loc='best'): chú thích tên đường

loc='best': Matplotlib tự chọn vị trí đẹp nhất để không che biểu đồ.
