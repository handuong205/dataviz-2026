1.groupby()

Khi nào dùng?

Dùng để nhóm dữ liệu theo một hoặc nhiều cột và thực hiện các phép tính như tổng, trung bình, đếm số lượng, lớn nhất, nhỏ nhất...

Syntax:

DataFrame.groupby(by)[column].agg(function)

Parameters quan trọng:

- by: cột dùng để nhóm
- column: cột cần xử lý
- agg(function): hàm tính toán (sum, mean, count, max, min...)

```python
students = pd.DataFrame({
    'Class': ['A', 'A', 'B', 'B', 'C'],
    'Name': ['An', 'Binh', 'Cuong', 'Dung', 'Ha'],
    'Score': [8, 9, 7, 6, 10]
})

result = students.groupby('Class')['Score'].mean()

print(result)
```

2.merge()

Khi nào dùng?

Dùng để nối nhiều DataFrame lại với nhau dựa trên cột chung.

Syntax:

pd.merge(df1, df2, on='column', how='inner')

Parameters quan trọng

- df1, df2: hai bảng cần nối
- on: cột chung
- how: kiểu nối
- inner
- left
- right
- outer

```python
students = pd.DataFrame({
    'StudentID': [1, 2, 3],
    'Name': ['An', 'Binh', 'Cuong']
})

scores = pd.DataFrame({
    'StudentID': [1, 2, 3],
    'Math': [9, 8, 7]
})

result = pd.merge(students, scores, on='StudentID')

print(result)
```

3. pivot_table()
Khi nào dùng?

Dùng để tạo bảng thống kê tổng hợp giống Pivot Table trong Excel.

Syntax:

pd.pivot_table(data, values, index, columns, aggfunc)

Parameters quan trọng
- data: DataFrame
- values: cột cần tính toán
- index: hàng
- columns: cột
- aggfunc: hàm thống kê

```python
sales = pd.DataFrame({
    'Region': ['North', 'North', 'South', 'South'],
    'Product': ['Laptop', 'Phone', 'Laptop', 'Phone'],
    'Revenue': [1000, 500, 1200, 700]
})

result = pd.pivot_table(
    sales,
    values='Revenue',
    index='Region',
    columns='Product',
    aggfunc='sum'
)

print(result)
```
